```python
seed = 124

numpy.random.seed(seed)
sample = numpy.random.choice(numpy.arange(len(X_test)), 100)

# print(sample)

X_sample = Xb.iloc[sample]
Y_sample = Yb.iloc[sample]

Yp = pandas.DataFrame(modelo.predict(X_sample))

figure, axes = pyplot.subplots(10, 10, figsize=(5, 5))

corrects = 0
fails = 0
total = 0

for i in range(10):
    for j in range(10):
        total = total + 1

        k = i * 10 + j
        indice_k = X_sample.index[k]

        digito_correcto = Y_sample.iloc[k].argmax()
        digito_prediccion = Yp.iloc[k].argmax()

        if digito_correcto == digito_prediccion:
            axes[i][j].imshow(X_sample.iloc[k].values.reshape(28, 28) / 255.0, cmap="Blues")
            corrects += 1
        else:
            axes[i][j].imshow(X_sample.iloc[k].values.reshape(28, 28) / 255.0, cmap="Reds")
            print(f"IMAGEN {k} / INDICE {indice_k}")
            print(f"DIGITO CORRECTO:   {digito_correcto}")
            print(f"DIGITO PREDICCION: {digito_prediccion}")
            print("=" * 40)
            fails += 1

pyplot.show()

print(f"SEED:     {seed}")
print("-" * 20)
print(f"TOTAL:    {total}")
print(f"CORRECTS: {corrects}")
print(f"FAILS:    {fails}")
```

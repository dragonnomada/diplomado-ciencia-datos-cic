```python
import pandas
import numpy
from matplotlib import pyplot
import seaborn

from PIL import Image
```

```python
catdogs = pandas.read_csv("catdogs.csv")

catdogs
```

```python
catdogs["PATH"]
```

```python
X = numpy.vstack(
    catdogs["PATH"].apply(
        lambda path: numpy.ascontiguousarray(Image.open(path).convert("L")).reshape(64 * 64))
        .apply(lambda x: (255 - x) / 255).values)

X
```

```python
catdogs["TYPE"]
```

```python
y1 = (catdogs["TYPE"] == "Cat").astype(int)
y2 = (catdogs["TYPE"] == "Dog").astype(int)

Y = pandas.DataFrame({
    "y1": y1,
    "y2": y2,
})

Y
```

```python
! pip install scikit-learn
```

```python
from sklearn.model_selection import train_test_split

Xa, Xb, Ya, Yb = train_test_split(X, Y, test_size=3000, random_state=123)
```

```python
Xa
```

```python
import keras

modelo = keras.Sequential([
    keras.Input(shape=(64 * 64,)),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(2, activation="sigmoid"),
])

modelo.summary()
```

```python
modelo.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
historial = modelo.fit(Xa, Ya, epochs=100)
```

```python
modelo.evaluate(Xb, Yb)
```

## Entrenamiento con una Red Neuronal Convolucional

```python
import keras

modelo = keras.Sequential([
    keras.Input(shape=(64, 64, 1)),
    keras.layers.Conv2D(32, 5, strides=2, activation="relu"),
    keras.layers.Conv2D(32, 3, strides=2, activation="relu"),
    keras.layers.MaxPool2D(5),
    keras.layers.Flatten(),
    keras.layers.Dense(2, activation="sigmoid"),
])

modelo.summary()
```

```python
modelo.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
X = numpy.vstack(
    catdogs["PATH"].apply(
        lambda path: numpy.ascontiguousarray(Image.open(path).convert("L")).reshape(64, 64, 1))
        .apply(lambda x: (255 - x) / 255).values).reshape(len(catdogs["PATH"]), 64, 64, 1)

X
```

```python
X.shape
```

```python
from sklearn.model_selection import train_test_split

Xa, Xb, Ya, Yb = train_test_split(X, Y, test_size=3000, random_state=123)
```

```python
historial = modelo.fit(Xa, Ya, epochs=35)
```

```python
modelo.evaluate(Xb, Yb)
```

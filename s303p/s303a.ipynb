# Procesamiento de imágenes de perros y gatos

1. Explorar las imágenes de gatos y perros y guardar sus rutas
2. Abrir las imágenes y normalizarlas a escala de grises, centradas de 28x28 pixeles
3. Guardar las imágenes normalizadas como `cat_0.png`, `cat_1.png`, ..., `cat_12499.png` y lo mismo para `dog_0.png`, `dog_1.png`, ..., `dog_12499.png`

```python
import pandas

import os
import re
```

```python
cats = []

for root, subs, files in os.walk("../../Cat"):
    for file in files:
        for match in re.finditer(r"(\d+)\.jpg", file):
            id = int(match.group(1))
            # print(file, id)
            cats.append((id, file))

cats = pandas.DataFrame(cats, columns=["ID", "PATH"])

cats["TYPE"] = "Cat"

cats = cats.sort_values("ID").reset_index().drop("index", axis=1)

cats
```

```python
dogs = []

for root, subs, files in os.walk("../../Dog"):
    for file in files:
        for match in re.finditer(r"(\d+)\.jpg", file):
            id = int(match.group(1))
            # print(file, id)
            dogs.append((id, file))

dogs = pandas.DataFrame(dogs, columns=["ID", "PATH"])

dogs["TYPE"] = "Dog"

dogs = dogs.sort_values("ID").reset_index().drop("index", axis=1)

dogs
```

```python
catdogs = pandas.concat([cats, dogs]).reset_index().drop("index", axis=1)

catdogs
```

```python
from PIL import Image
import numpy

# from matplotlib import pyplot

# figure, axes = pyplot.subplots(1, 2, figsize=(10, 10))

correctas = []

for (index, id, path, type) in catdogs.itertuples():
    if index % 1000 == 0:
        print(index, id, path, type)

    try:
        image = Image.open(f"../../{type}/{path}")
        image = image.convert("L")
        image = numpy.ascontiguousarray(image)

        height, width = image.shape

        L = max(width, height)
        dx = L - width
        dy = L - height

        # print(width, height, L, dx, dy)

        image2 = numpy.zeros((L, L)) + 255

        image2[int(dy/2):int(dy/2)+height, int(dx/2):int(dx/2)+width] = image

        # axes[0].imshow(255 - image, cmap="Grays")
        # axes[1].imshow(255 - image2, cmap="Grays")

        Image.fromarray(image2).resize((64, 64)).convert("RGB").save(f"../../catdogs/{type}_{id}.png")

        correctas.append((id, type, f"../../catdogs/{type}_{id}.png", f"../../{type}/{path}"))
    except Exception as e:
        print("ERROR:", id)

    # break

# pyplot.show()

correctas = pandas.DataFrame(correctas, columns=["ID", "TYPE", "PATH", "ORIGINAL_PATH"])

correctas.to_csv("catdogs.csv", index=False)

correctas
```
